```python
# Корпус из 4 документов
corpus=["oster-m1.txt", "oster-m2.txt", "oster-ph1.txt", "oster-ph2.txt"]
N=len(corpus) # Количество документов в корпусе
```

```python
import re
from nltk.corpus import stopwords
stopwords_ru = stopwords.words('russian') # загружаем список стоп-слов для русского языка
```

```python
# аргумент - имя файла с текстом, разделяем на слова, удаляем стоп-слова, возвращаем список слов
def lexemes(namefile):
    # 1. Разбираем текст из файла на слова:
    file=open(namefile,"r",encoding="utf-8")
    data=file.read()    
    file.close()
    #words=data.split()
    #
    import re
    #words=re.split(r'[-\s.,:;!?)()—»«"]+',data)
    words=re.split(r'[^а-яА-Яa-zA-Z]+',data)
    # 2. Удаляем стоп-слова, приводим к нижнему регистру:        
    listWords = [x.lower() for x in words if x.lower() not in stopwords_ru]         
    return listWords
```

```python
# аргумент - имя файла с текстом, разделяем на слова, удаляем стоп-слова, приводим к нормальной форме, возвращаем список лемм
def lemmes(namefile):    
    # 1. Разбираем текст из файла на слова:
    file=open(namefile,"r",encoding="utf-8")
    data=file.read()    
    file.close()    
    #
    import re    
    words=re.split(r'[^а-яА-Яa-zA-Z]+',data)
    # 2. Удаляем стоп-слова, приводим к нижнему регистру:        
    listWords = [x.lower() for x in words if x.lower() not in stopwords_ru] 
    #
    import pymorphy2
    morph = pymorphy2.MorphAnalyzer()
    lemmedWords = [morph.parse(x)[0].normal_form for x in listWords]
    return lemmedWords
```

```python
# 1. Создаем двумерный список всех слов, входящих в документы корпуса
#   [[слова документа_0], [слова документа_1], ..., [слова документа_N-1]]
corpusWords=[]
for d in corpus:
    #listwords_d=lexemes(d)# список слов документа d
    listwords_d=lemmes(d)# список слов-лемм документа d
    # добавляем полученный список в общий список    
    corpusWords.append(listwords_d)
print(len(corpusWords[3]))
print(len(set(corpusWords[3])))
# Документ_0: слов 9004, уникальных слов 3930, лемм 2337
# Документ_1: слов 7698, уникальных слов 3609,лемм 2251
# Документ_2: слов 10093, уникальных слов 4758,лемм 2848
# Документ_3: слов 2921, уникальных слов 1633,лемм 1147
```

```python
# Создаем Лексикон - набор уникальных слов в объединении множеств слов корпуса документов
#   формируем список list_all - объединенный список слов всех документов корпуса.
#   corpusWords - двумерный список слов документов корпуса, созданный на шаге 1
#  
list_all=sum(corpusWords,[]) # второй аргумент - [] - начальное значение
lexicon=sorted(set(list_all))
#print(lexicon)
print(len(lexicon)) # число уникальных слов в корпусе документов
# 7982 слов
# 4378 лемм
```

```python
# Мешок слов - BoW (bag-of-words)
def compute_bow(list_words):
    # создаем словарь слов и их встречаемости в  документе
    uniqeWordsList=set(list_words) # ?
    dictWords=dict.fromkeys(uniqeWordsList,0)
    for word in list_words:
        dictWords[word] +=1     
    return dictWords

#
# Создаем мешки слов документов корпуса
listBoWs=[] # список "мешков слов" документов
for i in range(N): # N - количество документов в корпусе
    # цикл по документам корпуса,
    # точнее по спискам слов документов. Эти списки получены на шаге 1.
    bow_i=compute_bow(corpusWords[i])
    listBoWs.append(bow_i)
# Теперь listBoWs[0] - мешок слов документа corpus[0], listBoWs[1] - мешок слов документа corpus[1], ...
#print(len(listBoWs[0]))
#print(len(listBoWs[1]))
print(listBoWs[1])
```

```python
# Создаем список, содержащий векторы для каждого документа корпуса
# вектор документа имеет размер, равный размеру lexicon.
# Компонент с номером m вектора документа d равен 0, если m-ое "слово" лексикона  отсутствует в документе d 
#                   и  количеству упоминаний этого слова в d, если слово встречается в d.
#
vect_docs=[] 
for m in range(N):
    vect=[]
    for word in lexicon:
        if word in set(listBoWs[m]):
            vect.append(listBoWs[m][word])
        else:
            vect.append(0)           
    vect_docs.append(vect)  
#print(vect_docs[0])
```

```python
def cosine_sim(vec1, vec2):
    import math     
    dot_prod = 0
    for i, v in enumerate(vec1):
        dot_prod += v * vec2[i]        
    mag_1 = math.sqrt(sum([x**2 for x in vec1]))
    mag_2 = math.sqrt(sum([x**2 for x in vec2]))    
    return dot_prod / (mag_1 * mag_2)
```

```python
import math
d_0_1 = cosine_sim(vect_docs[0], vect_docs[1])
print('cos =',d_0_1)
print('Угол между векторами: ', math.degrees(math.acos(d_0_1)))
# d_0 -"oster-m1.txt" d_1 "oster-m2.txt"
```

```python
# Ещё один способ частотной векторизации
#  вычисление словаря {слово:TF этого слова} в списке wordDict
#    TF(t,d)= tf(t,d)/число терминов в d
def compute_TF(wordDict):
    tfDict = {}
    WordsCount = len(wordDict)
    for word, count in wordDict.items():
        tfDict[word] = count / float(WordsCount)
    return tfDict
```

```python
# Создаем мешки слов документов корпуса
listTF=[] # список "мешков слов" документов
for i in range(N): # N - количество документов в корпусе
    # цикл по документам корпуса
    bow_i=compute_TF(listBoWs[i])
    listTF.append(bow_i)
# Теперь listBoWs[0] - мешок слов документа corpus[0], listBoWs[1] - мешок слов документа corpus[1], ...
#print(listTF[0]) 
```


Векторизация TF-IDF = TF x IDF.

```python
def comp_idf(lexicon,corpusWords):
    import math
    N = len(corpusWords)
    idf_lexicon={k:0 for k in lexicon} # 1. начальный словарь - {'слово1':0, 'слово2':0, ... }
    listCorpusSet=[] # 2. список множеств уникальных слов документов корпуса
    for d in corpusWords: # 3. или for i in range(N):
        l_s=set(d) # 4. или l_s=set(corpusWords[i])
        listCorpusSet.append(l_s) # 5.
    for key in idf_lexicon: # for key in lexicon
       # проходим по всем документам корпуса
       # если слово содержится в очередном документе, добавляем 1 в value
        for s in listCorpusSet: # или for i in range(N):
            # если слово key содержится в множестве слов очередного документа
           if key in s: # или if key in set(corpusWords[i]). И можно без операций 2 - 5.
               idf_lexicon[key] += 1
    #
    for word, val in idf_lexicon.items():
        idf_lexicon[word] = math.log(N / float(val)) 
    return idf_lexicon
```

```python
def compTFIDF(tf_bow, idfs):
    # 
    tfidf = {}
    for word, val in tf_bow.items():
        tfidf[word] = val * idfs[word]
    return tfidf
```

```python
IDF_lexicon = comp_idf(lexicon,corpusWords)
#print(IDF_lexicon)
```

```python
# двумерный список, содержащий значения tf-idf слов документов корпуса, например,
# tfidf_corpus[0] - список, содержащий значения tf-idf слов первого документ
tfidf_corpus= [] 
for i in range(N):
    tmp=compTFIDF(listBoWs[i], IDF_lexicon)
    tfidf_corpus.append(tmp)
#print(tfidf_corpus[0]) # список значений tf-idf первого документа
```

```python
# И снова создаем список, содержащий векторы tfidf для каждого документа корпуса
vect_docs_tfidf=[] 
for m in range(N):
    vect=[]
    for word in lexicon:
        if word in set(tfidf_corpus[m]):
            vect.append(tfidf_corpus[m][word])
        else:
            vect.append(0)           
    vect_docs_tfidf.append(vect)  
#print(vect_docs_[0])
```

```python
# определение векторов TFIDF для нескольких терминов
terms=["задача","длина","ширина"] # термины
vect_terms_TFIDF=[] 
for m in range(N):
    vect=[]
    for word in terms:
        if word in set(corpusWords[m]):
            vect.append(tfidf_corpus[m][word])
        else:
            vect.append(0)           
    vect_terms_TFIDF.append(vect)  
#
print(vect_terms_TFIDF)
print(vect_terms_TFIDF[3])
print("         задача         длина      ширина")
for m in range(N):   
    print("Док {0}:   {1:.3f}          {2:.3f}      {3:.3f}".format(m,vect_terms_TFIDF[m][0],vect_terms_TFIDF[m][1],vect_terms_TFIDF[m][2]))
```
