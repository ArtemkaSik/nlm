# MorphVocab

```python
from natasha import MorphVocab, Doc, Segmenter, DatesExtractor, NewsEmbedding, NewsSyntaxParser, NewsMorphTagger
```

```python
text3 = 'Глокая куздра штеко будланула бокра и курдячит бокрёнка'
```

```python
doc = Doc(text3)
segmenter = Segmenter()
emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)

doc.segment(segmenter)  # для обращения к doc.tokens
doc.tag_morph(morph_tagger)  # для использования token.lemmatize

morph_vocab = MorphVocab()

for token in doc.tokens:
    token.lemmatize(morph_vocab)
{d.text: d.lemma for d in doc.tokens}
```

**Токенизация и лемматизация** (приведение к нормальной форме) полученных токенов

# Синтаксический анализ 

```python
from natasha import Segmenter, NewsEmbedding, NewsSyntaxParser, Doc
```

```python
text8 = """
Преподаватель предложил прийти на зачет во вторник.
Преподаватель предложил во вторник всё-таки прийти на зачет.
Мать любит дочь.
Ему нужно отдать долг.
"""
```

Синтаксическая омонимия (наличие у одного предложения двух или более правильных синтаксических анализов)

```python
doc = Doc(text8)
segmenter = Segmenter()
emb = NewsEmbedding()
syntax_parser = NewsSyntaxParser(emb)
doc.segment(segmenter)
doc.parse_syntax(syntax_parser)
for sent in doc.sents:
    sent.syntax.print()
    print()
```

В первом предложении программа решила, что преподаватель предложил **прийти во вторник**, а во втором предложении, что **предложил во вторник** прийти на зачёт.

Так как морфологический модуль в `natasha` реализован с помощью `pymorphy2`, то можно использовать ту же функцию `make_agree_with_number`. Но, как оказалось, в `natasha` существует баг, и при её использовании появляется ошибка.

```python
# Ссылка на исходный код реализации метода make_agree_with_number
# https://github.com/pymorphy2/pymorphy2/blob/92d546f042ff14601376d3646242908d5ab786c1/pymorphy2/analyzer.py#L38

print(word.tag.numeral_agreement_grammemes(5))
```

Оказалось, что `make_agree_with_number` реализован через метод `numeral_agreement_grammemes`  атрибута `tag` у слова. Я использовал этот метод и получил грамеммы, которые нужны для соответствия слова "кочерга" числу 5.

```python
# Ссылка на реализацию маппинга тегов слов
# https://github.com/natasha/natasha/blob/7fe3712f956622733145c52eb5c2942a2afdf0bb/natasha/morph/vocab.py#L101

# Ссылка на массив использующихся тегов в natasha в виде [тип тега, ключ словаря, значение словаря]
# https://github.com/natasha/natasha/blob/7fe3712f956622733145c52eb5c2942a2afdf0bb/natasha/morph/vocab.py#L40

result = word.inflect({"Gen", "Plur"}).word
print(textA.format(result))
```

Проблема в том, что `natasha` заточена под трёхбуквенный формат граммем с заглавной буквы, а метод из `pymorphy2` возвращает грамеммы в четырёхбуквенном формате со строчной.

После использования метода `inflect` у слова, по аналогии с внутренней реализацией `make_agree_with_number`, с переданными грамеммами в исправленном формате, слово, наконец, приняло правильную форму.
