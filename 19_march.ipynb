### 1) Когда TF и IDF достигают минимального значения?

**TF (Term Frequency)**
*   **Минимальное значение:** `0`
*   **Условие:** Слово **не встречается** в конкретном документе.
*   **Логика:** Если счетчик слов равен 0, то и частота равна 0.

**IDF (Inverse Document Frequency)**
*   **Минимальное значение:** `0`
*   **Условие:** Слово встречается **во всех документах** коллекции.
*   **Формула:** $\log(\frac{N}{N}) = \log(1) = 0$
*   **Смысл:** Если слово есть везде (например, предлоги «и», «в»), оно не помогает отличить один документ от другого, поэтому его вес обесценивается до минимума.

### 2) Может ли TF и IDF быть больше 1?

**TF (Term Frequency)**
*   **Ответ :** **Да**.
    *   TF — это просто количество вхождений слова в документ. Нет никакого ограничения сверху.
    
**IDF (Inverse Document Frequency)**
*   **Ответ:** **Да, часто бывает больше 1**.
*   **Условие:** Когда слово встречается **редко** (в меньшинстве документов).
*   **Пример:**
    *   В коллекции 100 документов ($N=100$).
    *   Слово встретилось только в 1 документе ($n_t=1$).
    *   Расчет: $\log(\frac{100}{1}) = \log(100)$.
    *   При десятичном логарифме результат равен **2**.
    *   При натуральном логарифме результат равен **~4.6**.
*   **Смысл:** Редкие, уникальные слова получают высокий вес (>1), так как они наиболее информативны для классификации документа.
