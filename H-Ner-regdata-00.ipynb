```python
file=open("dates_test_sample.txt","r", encoding="utf-8")
text=file.read()
file.close()
print(text)
```

БЫЛО

```python
import re
eu = r'((([0123]?\d)[-/]([01]?\d))([-/]([0123]\d)?\d\d)?)'# Регулярное выражение для европейских дат. 
# Обработка естественного языка в действии, стр. 412
# Добавлена \b в начале и конце. Использованы незахватывающие группы (?:...), чтобы findall возвращал только дату целиком, а не кортеж из кусочков.
# \b - граница слова
dmy = re.findall(eu,text)
print('БЫЛО:')
dmy
```

СТАЛО

```python
import re
# eu = r'((([0123]?\d)[-/]([01]?\d))([-/]([0123]\d)?\d\d)?)'# Регулярное выражение для европейских дат. 
# Обработка естественного языка в действии, стр. 412
eu = r'\b((?:[0123]?\d)[-/](?:[01]?\d)(?:[-/](?:(?:[0123]\d)?\d\d))?)\b'
# Добавлена \b в начале и конце. Использованы незахватывающие группы (?:...), чтобы findall возвращал только дату целиком, а не кортеж из кусочков.
# \b - граница слова
dmy = re.findall(eu,text)
print('СТАЛО')
dmy
```
